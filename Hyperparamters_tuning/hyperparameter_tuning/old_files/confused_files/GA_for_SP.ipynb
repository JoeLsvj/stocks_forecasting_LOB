## Long Short-Term Memory (LSTM)

Long Short-Term Memory (LSTM) networks are a type of recurrent neural network (RNN) that are well-suited to learn from and make predictions on sequential data. Unlike standard RNNs, LSTMs are capable of learning long-term dependencies, which is achieved by their unique structure that includes memory cells and three types of gates: input, output, and forget gates. These gates regulate the flow of information, allowing the network to retain important information over long periods.

LSTMs are particularly useful in applications such as time series forecasting, natural language processing (NLP), and speech recognition.

- Ability to remember long-term dependencies.
- Mitigation of the vanishing gradient problem through gating mechanisms.
- Effective for tasks involving sequential data.

## Convolutional Neural Network (CNN)

Convolutional Neural Networks (CNNs) are a class of deep neural networks commonly used for analyzing visual data. They are specifically designed to process pixel data and are thus widely used in image recognition and computer vision tasks. CNNs are composed of convolutional layers, pooling layers, and fully connected layers. The convolutional layers apply filters to the input image to create feature maps, which are then downsampled by the pooling layers to reduce dimensionality and computation requirements.

CNNs have proven highly effective in tasks such as image classification, object detection, and image segmentation.

- Convolutional layers for automatic feature extraction.
- Pooling layers for dimensionality reduction.
- Fully connected layers for classification.

# eXtreme Gradient Boosting (XGBoost)

eXtreme Gradient Boosting (XGBoost) is an optimized distributed gradient boosting library designed to be highly efficient, flexible, and portable. It implements machine learning algorithms under the Gradient Boosting framework and is known for its performance and speed. XGBoost provides a parallel tree boosting that solves many data science problems in a fast and accurate way.

XGBoost is used for supervised learning tasks such as regression, classification, and ranking.

- High performance and speed.
- Parallel processing capability.
- Regularization to prevent overfitting.

# Light Gradient Boosting Machine (LightGBM)

Light Gradient Boosting Machine (LightGBM) is a gradient boosting framework that uses tree-based learning algorithms. It is designed to be distributed and efficient, with the ability to handle large-scale data. LightGBM provides faster training speed and higher efficiency, lower memory usage, and better accuracy than other gradient boosting algorithms.

LightGBM is well-suited for large datasets and can be used for various machine learning tasks, including classification, regression, and ranking.

- Efficient handling of large datasets.
- Faster training and higher efficiency.
- Reduced memory usage.
